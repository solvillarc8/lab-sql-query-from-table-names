# SQL query from table names

In This notebook we are going to test if using just the name of the table, and a shord definition of its contect we can use a model like GTP3.5-Turbo to select which tables are necessary to create a SQL Order to answer the user petition.

```python
from openai import OpenAI
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())

OPENAI_API_KEY  = os.getenv('OPENAI_API_KEY')
```

```python
#Functio to call the model.
def return_OAI(user_message):
    client = OpenAI(
    # This is the default and can be omitted
    api_key=OPENAI_API_KEY,
)
    context = []
    context.append({'role':'system', "content": user_message})

    response = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=context,
            temperature=0,
        )

    return (response.choices[0].message.content)
```

```python
#Definition of the tables.
import pandas as pd

# Table and definitions sample
data = {'table': ['employees', 'salary', 'studies'], # ENTER A TABLE COLUMNS HERE,
        'definition': [
            'Employee information including employee ID, name, department, hire date, and job title',
            'Salary details for each employee per year, including base salary, bonuses, and currency',
            'Educational studies of each employee: name of the institution, type of studies, level (bachelor, master, PhD), and graduation year'
        ]
       }# ENTER A TABLE DEFINITATIONS HERE
df = pd.DataFrame(data)
print(df)
```

```python
text_tables = '\n'.join([f"{row['table']}: {row['definition']}" for index, row in df.iterrows()])
```

```python
print(text_tables)
```

```python
prompt_question_tables = """
Given the following tables and their content definitions,
###Tables
{tables}

Tell me which tables would be necessary to query with SQL to address the user's question below.
Return the table names in a json format.
###User Questyion:
{question}
"""
```

```python
#Creating the prompt, with the user questions and the tables definitions.
pqt1 = prompt_question_tables.format(
    tables=text_tables,
    question="What is the average salary per department for the current year?"
)
```

```python
print(return_OAI(pqt1))
```

```python
pqt3 = prompt_question_tables.format(
    tables=text_tables,
    question="Show me all employees with a Master's or PhD degree along with their salary history and years of experience."
)
```

```python
print(return_OAI(pqt3))
```

## Findings Report: SQL Table Selection with LLMs

### Objective
The goal was to test whether GPT-3.5-Turbo could correctly identify which database tables are needed to answer a natural language question — using only table names and short descriptions, without seeing any schema or data.

### What Was Tested
Three tables were defined:
- **employees** – employee info (ID, name, department, hire date, title)
- **salary** – yearly salary details per employee
- **studies** – educational background (institution, level, graduation year)

### Key Learnings
1. **LLMs can act as a routing layer** — even with minimal metadata (just a name and one-line definition), the model makes reasonable decisions about which tables are relevant.
2. **Output format is not deterministic** — without explicit format enforcement (e.g., *"return ONLY a JSON array of table names, no extra keys"*), the model's output structure can vary, which breaks downstream parsing.
